```python
# Load all datasets
orders    = pd.read_csv(r'C:\Users\yashy\orders.csv')
customers = pd.read_csv(r'C:\Users\yashy\customers.csv')
products  = pd.read_csv(r'C:\Users\yashy\products.csv')
returns   = pd.read_csv(r'C:\Users\yashy\returns.csv')

print("=== ORDERS SHAPE ===")
print(f"Rows: {orders.shape[0]}, Columns: {orders.shape[1]}")

print("\n=== COLUMN NAMES ===")
print(list(orders.columns))

print("\n=== DATA TYPES ===")
print(orders.dtypes)

print("\n=== MISSING VALUES ===")
print(orders.isnull().sum())

print("\n=== FIRST 3 ROWS ===")
print(orders.head(3))
```

```python
# Fix date columns
orders['order_date'] = pd.to_datetime(orders['order_date'])
orders['ship_date']  = pd.to_datetime(orders['ship_date'])
returns['return_date'] = pd.to_datetime(returns['return_date'])

# Extract useful date features
orders['order_year']  = orders['order_date'].dt.year
orders['order_month'] = orders['order_date'].dt.month
orders['order_month_name'] = orders['order_date'].dt.strftime('%b')
orders['ship_days']   = (orders['ship_date'] - orders['order_date']).dt.days

# Verify
print("=== UPDATED DATA TYPES ===")
print(orders[['order_date', 'ship_date', 'order_year', 
              'order_month', 'ship_days']].dtypes)

print("\n=== NEW COLUMNS SAMPLE ===")
print(orders[['order_date', 'order_year', 
              'order_month', 'order_month_name', 
              'ship_days']].head(5))

print("\n=== SHIPPING DAYS STATS ===")
print(orders['ship_days'].describe())
```

```python
# Merge orders with products and customers
df = orders.merge(products, on='product_id', how='left') \
           .merge(customers[['customer_id','customer_name',
                              'city','segment']], 
                  on='customer_id', how='left')

# Add return flag
df['is_returned'] = df['order_id'].isin(returns['order_id']).astype(int)

# Final check
print("=== MERGED DATASET ===")
print(f"Shape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
print(f"\nReturned Orders: {df['is_returned'].sum()}")
print(f"Return Rate: {df['is_returned'].mean()*100:.2f}%")

print("\n=== SAMPLE ROW ===")
print(df[['order_id','order_date','customer_name',
          'product_name','category','region',
          'sales_amount','profit','is_returned']].head(3))
```

```python
print("=== BUSINESS SUMMARY ===")
print(f"Total Orders      : {len(df):,}")
print(f"Total Revenue     : ₹{df['sales_amount'].sum():,.2f}")
print(f"Total Profit      : ₹{df['profit'].sum():,.2f}")
print(f"Profit Margin     : {(df['profit'].sum()/df['sales_amount'].sum()*100):.2f}%")
print(f"Avg Order Value   : ₹{df['sales_amount'].mean():,.2f}")
print(f"Avg Profit/Order  : ₹{df['profit'].mean():,.2f}")
print(f"Total Customers   : {df['customer_id'].nunique():,}")
print(f"Total Products    : {df['product_name'].nunique():,}")
print(f"Return Rate       : {df['is_returned'].mean()*100:.2f}%")
print(f"Avg Shipping Days : {df['ship_days'].mean():.2f}")

print("\n=== REVENUE BY CATEGORY ===")
category_summary = df.groupby('category').agg(
    total_orders   = ('order_id','count'),
    total_revenue  = ('sales_amount','sum'),
    total_profit   = ('profit','sum'),
    avg_margin     = ('profit', lambda x: (x.sum()/df.loc[x.index,'sales_amount'].sum()*100))
).round(2)
print(category_summary.sort_values('total_revenue', ascending=False))

print("\n=== REVENUE BY REGION ===")
region_summary = df.groupby('region').agg(
    total_orders  = ('order_id','count'),
    total_revenue = ('sales_amount','sum'),
    total_profit  = ('profit','sum')
).round(2)
print(region_summary.sort_values('total_revenue', ascending=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('RetailCo — Category Performance', 
             fontsize=16, fontweight='bold')

# Chart 1 — Revenue by Category
category_rev = df.groupby('category')['sales_amount'].sum().sort_values(ascending=False)
axes[0].bar(category_rev.index, category_rev.values, 
            color=['#2196F3', '#4CAF50', '#FF9800'])
axes[0].set_title('Revenue by Category', fontweight='bold')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Revenue (₹)')
axes[0].yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.1f}M'))
for i, v in enumerate(category_rev.values):
    axes[0].text(i, v + 100000, f'₹{v/1e6:.1f}M', 
                ha='center', fontweight='bold')

# Chart 2 — Profit Margin by Category
category_margin = (df.groupby('category')['profit'].sum() / 
                   df.groupby('category')['sales_amount'].sum() * 100).round(2)
axes[1].bar(category_margin.index, category_margin.values,
            color=['#E91E63', '#9C27B0', '#00BCD4'])
axes[1].set_title('Profit Margin % by Category', fontweight='bold')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Margin %')
for i, v in enumerate(category_margin.values):
    axes[1].text(i, v + 0.3, f'{v}%', 
                ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\category_performance.png', 
            dpi=150, bbox_inches='tight')
plt.show()
print("✅ Chart saved!")
```

```python
fig, axes = plt.subplots(2, 1, figsize=(14, 10))
fig.suptitle('RetailCo — Sales Trend Analysis', 
             fontsize=16, fontweight='bold')

# Monthly revenue
monthly = df.groupby(['order_year', 'order_month']).agg(
    revenue = ('sales_amount', 'sum'),
    profit  = ('profit', 'sum'),
    orders  = ('order_id', 'count')
).reset_index()
monthly['period'] = monthly['order_year'].astype(str) + '-' + \
                    monthly['order_month'].astype(str).str.zfill(2)

# Chart 1 — Revenue vs Profit Trend
axes[0].plot(monthly['period'], monthly['revenue'], 
             marker='o', color='#2196F3', linewidth=2, label='Revenue')
axes[0].plot(monthly['period'], monthly['profit'],  
             marker='s', color='#4CAF50', linewidth=2, label='Profit')
axes[0].set_title('Monthly Revenue vs Profit', fontweight='bold')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Amount (₹)')
axes[0].legend()
axes[0].tick_params(axis='x', rotation=45)
axes[0].yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.1f}M'))

# Chart 2 — Monthly Orders
axes[1].bar(monthly['period'], monthly['orders'],
            color='#FF9800', alpha=0.8)
axes[1].set_title('Monthly Order Volume', fontweight='bold')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Number of Orders')
axes[1].tick_params(axis='x', rotation=45)
for i, v in enumerate(monthly['orders']):
    axes[1].text(i, v + 0.5, str(v), ha='center', fontsize=8)

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\monthly_trend.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("✅ Trend chart saved!")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 7))
fig.suptitle('RetailCo — Product Performance', 
             fontsize=16, fontweight='bold')

# Product revenue
product_rev = df.groupby('product_name')['sales_amount'].sum()\
                .sort_values(ascending=True)
product_profit = df.groupby('product_name')['profit'].sum()\
                   .sort_values(ascending=True)

# Chart 1 — Revenue by Product
axes[0].barh(product_rev.index, product_rev.values, color='#2196F3')
axes[0].set_title('Revenue by Product', fontweight='bold')
axes[0].set_xlabel('Revenue (₹)')
axes[0].xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.1f}M'))
for i, v in enumerate(product_rev.values):
    axes[0].text(v + 50000, i, f'₹{v/1e6:.1f}M', 
                va='center', fontsize=9)

# Chart 2 — Profit Margin by Product
product_margin = (df.groupby('product_name')['profit'].sum() /
                  df.groupby('product_name')['sales_amount'].sum() * 100)\
                  .round(2).sort_values(ascending=True)
colors = ['#F44336' if x < 35 else '#4CAF50' for x in product_margin.values]
axes[1].barh(product_margin.index, product_margin.values, color=colors)
axes[1].set_title('Profit Margin % by Product\n(Red = Below 35%)', 
                  fontweight='bold')
axes[1].set_xlabel('Margin %')
for i, v in enumerate(product_margin.values):
    axes[1].text(v + 0.2, i, f'{v}%', va='center', fontsize=9)

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\product_performance.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("✅ Product chart saved!")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('RetailCo — Regional Analysis', 
             fontsize=16, fontweight='bold')

# Pivot for heatmap — Region vs Category
heatmap_data = df.pivot_table(
    values='profit',
    index='region',
    columns='category',
    aggfunc='sum'
).round(0)

# Chart 1 — Heatmap
sns.heatmap(heatmap_data, 
            annot=True, 
            fmt='.0f',
            cmap='YlOrRd',
            ax=axes[0],
            linewidths=0.5)
axes[0].set_title('Profit Heatmap\nRegion vs Category', 
                  fontweight='bold')

# Chart 2 — Regional Revenue Bar
region_data = df.groupby('region').agg(
    revenue = ('sales_amount','sum'),
    profit  = ('profit','sum')
).reset_index()

x = range(len(region_data))
width = 0.35
axes[1].bar([i - width/2 for i in x], region_data['revenue'],
            width, label='Revenue', color='#2196F3', alpha=0.8)
axes[1].bar([i + width/2 for i in x], region_data['profit'],
            width, label='Profit', color='#4CAF50', alpha=0.8)
axes[1].set_title('Revenue vs Profit by Region', fontweight='bold')
axes[1].set_xticks(x)
axes[1].set_xticklabels(region_data['region'])
axes[1].set_ylabel('Amount (₹)')
axes[1].legend()
axes[1].yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.1f}M'))

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\regional_analysis.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("✅ Regional chart saved!")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('RetailCo — Correlation Analysis', 
             fontsize=16, fontweight='bold')

# Correlation matrix
numeric_cols = ['quantity', 'discount_pct', 
                'sales_amount', 'profit', 'ship_days']
corr_matrix = df[numeric_cols].corr().round(2)

# Chart 1 — Correlation Heatmap
sns.heatmap(corr_matrix,
            annot=True,
            cmap='coolwarm',
            center=0,
            ax=axes[0],
            linewidths=0.5,
            fmt='.2f')
axes[0].set_title('Correlation Matrix', fontweight='bold')

# Chart 2 — Discount vs Profit Scatter
axes[1].scatter(df['discount_pct'], df['profit'],
                alpha=0.4, color='#E91E63', edgecolors='white',
                linewidth=0.5)
axes[1].set_title('Discount % vs Profit', fontweight='bold')
axes[1].set_xlabel('Discount %')
axes[1].set_ylabel('Profit (₹)')
axes[1].yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e3:.0f}K'))

# Add trend line
z = np.polyfit(df['discount_pct'], df['profit'], 1)
p = np.poly1d(z)
x_line = np.linspace(df['discount_pct'].min(), 
                     df['discount_pct'].max(), 100)
axes[1].plot(x_line, p(x_line), 
             color='black', linewidth=2, 
             linestyle='--', label='Trend')
axes[1].legend()

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\correlation_analysis.png',
            dpi=150, bbox_inches='tight')
plt.show()

print("\n=== CORRELATION WITH PROFIT ===")
print(corr_matrix['profit'].sort_values(ascending=False))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('RetailCo — Customer Insights', 
             fontsize=16, fontweight='bold')

# Customer spending summary
customer_summary = df.groupby(['customer_id', 'segment']).agg(
    total_orders  = ('order_id', 'count'),
    total_spent   = ('sales_amount', 'sum'),
    total_profit  = ('profit', 'sum'),
    avg_order_val = ('sales_amount', 'mean')
).reset_index()

# Assign tiers
customer_summary['tier'] = pd.cut(
    customer_summary['total_spent'],
    bins=[0, 200000, 500000, float('inf')],
    labels=['Occasional', 'Regular', 'VIP']
)

# Chart 1 — Customer Tier Distribution
tier_counts = customer_summary['tier'].value_counts()
colors = ['#FFD700', '#C0C0C0', '#CD7F32']
axes[0,0].pie(tier_counts.values, 
              labels=tier_counts.index,
              colors=colors,
              autopct='%1.1f%%',
              startangle=90)
axes[0,0].set_title('Customer Tier Distribution', fontweight='bold')

# Chart 2 — Revenue by Segment
segment_rev = df.groupby('segment')['sales_amount'].sum()
axes[0,1].bar(segment_rev.index, segment_rev.values,
              color=['#2196F3', '#4CAF50', '#FF9800'])
axes[0,1].set_title('Revenue by Segment', fontweight='bold')
axes[0,1].set_ylabel('Revenue (₹)')
axes[0,1].yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.1f}M'))
for i, v in enumerate(segment_rev.values):
    axes[0,1].text(i, v + 100000, f'₹{v/1e6:.1f}M',
                   ha='center', fontweight='bold')

# Chart 3 — Orders by Payment Mode
payment = df['payment_mode'].value_counts()
axes[1,0].bar(payment.index, payment.values,
              color=['#E91E63','#9C27B0','#00BCD4','#FF5722'])
axes[1,0].set_title('Orders by Payment Mode', fontweight='bold')
axes[1,0].set_ylabel('Number of Orders')
for i, v in enumerate(payment.values):
    axes[1,0].text(i, v + 2, str(v), ha='center', fontweight='bold')

# Chart 4 — Segment vs Avg Order Value
seg_avg = customer_summary.groupby('segment')['avg_order_val'].mean()
axes[1,1].barh(seg_avg.index, seg_avg.values,
               color=['#2196F3', '#4CAF50', '#FF9800'])
axes[1,1].set_title('Avg Order Value by Segment', fontweight='bold')
axes[1,1].set_xlabel('Avg Order Value (₹)')
axes[1,1].xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e3:.0f}K'))
for i, v in enumerate(seg_avg.values):
    axes[1,1].text(v + 500, i, f'₹{v/1e3:.1f}K',
                   va='center', fontweight='bold')

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\customer_insights.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("✅ Customer insights chart saved!")
print(f"\nVIP Customers    : {(customer_summary['tier']=='VIP').sum()}")
print(f"Regular Customers: {(customer_summary['tier']=='Regular').sum()}")
print(f"Occasional       : {(customer_summary['tier']=='Occasional').sum()}")
```

```python
fig = plt.figure(figsize=(20, 12))
fig.suptitle('RetailCo — Executive Summary Dashboard', 
             fontsize=20, fontweight='bold', y=1.02)

# ── KPI Cards ──────────────────────────────────────────
kpis = [
    ('Total Revenue',   f"₹{df['sales_amount'].sum()/1e7:.2f} Cr", '#2196F3'),
    ('Total Profit',    f"₹{df['profit'].sum()/1e7:.2f} Cr",       '#4CAF50'),
    ('Profit Margin',   f"{df['profit'].sum()/df['sales_amount'].sum()*100:.1f}%", '#FF9800'),
    ('Total Orders',    f"{len(df):,}",                             '#E91E63'),
    ('Avg Order Value', f"₹{df['sales_amount'].mean():,.0f}",       '#9C27B0'),
    ('Return Rate',     f"{df['is_returned'].mean()*100:.1f}%",     '#F44336'),
]

for i, (label, value, color) in enumerate(kpis):
    ax = fig.add_subplot(3, 4, i+1)
    ax.set_facecolor(color)
    ax.text(0.5, 0.6, value, transform=ax.transAxes,
            fontsize=22, fontweight='bold', color='white',
            ha='center', va='center')
    ax.text(0.5, 0.2, label, transform=ax.transAxes,
            fontsize=11, color='white',
            ha='center', va='center')
    ax.set_xticks([])
    ax.set_yticks([])

# ── Revenue Trend ───────────────────────────────────────
ax_trend = fig.add_subplot(3, 4, (7, 8))
monthly_rev = df.groupby('period')['sales_amount'].sum() \
                if 'period' in df.columns else \
                df.groupby(df['order_date'].dt.to_period('M'))['sales_amount'].sum()
monthly_rev.plot(ax=ax_trend, color='#2196F3', linewidth=2, marker='o')
ax_trend.set_title('Monthly Revenue Trend', fontweight='bold')
ax_trend.tick_params(axis='x', rotation=45)
ax_trend.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.0f}M'))

# ── Product Revenue ─────────────────────────────────────
ax_prod = fig.add_subplot(3, 4, (9, 10))
prod_rev = df.groupby('product_name')['sales_amount'].sum()\
             .sort_values(ascending=True).tail(5)
ax_prod.barh(prod_rev.index, prod_rev.values, color='#4CAF50')
ax_prod.set_title('Top 5 Products by Revenue', fontweight='bold')
ax_prod.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.0f}M'))

# ── Region Profit ───────────────────────────────────────
ax_reg = fig.add_subplot(3, 4, (11, 12))
reg_profit = df.groupby('region')['profit'].sum().sort_values(ascending=False)
ax_reg.bar(reg_profit.index, reg_profit.values, color='#FF9800')
ax_reg.set_title('Profit by Region', fontweight='bold')
ax_reg.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, p: f'₹{x/1e6:.0f}M'))

plt.tight_layout()
plt.savefig(r'C:\Users\yashy\executive_dashboard.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("✅ Executive dashboard saved!")
```
